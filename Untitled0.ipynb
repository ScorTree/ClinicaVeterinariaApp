```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy.stats import norm
import os


nomArch = input("Archivo CSV (sin extensión): ")
df = pd.read_csv(nomArch+".csv")

# SE CREA UN DIRECTORIO PARA ALMACENAR LAS GRÁFICAS
directorio = 'Est-'+nomArch
if not os.path.exists(directorio):
    os.mkdir(directorio)

# SE CREA UN ARCHIVO TXT DONDE SE GUARDAN LOS DATOS DE LA ESTADÍSTICA DESCRIPTIVA
nomArchSalida = "Est-"+nomArch+".txt"
if os.path.exists(nomArchSalida):
    os.remove(nomArchSalida)
salida = f = open(nomArchSalida, 'w')


salida.write("Estadística descriptiva del archivo: "+str(nomArch)+'\n\n')

variables = df.columns # LEEMOS LOS METADATOS (variables)
tipos = df.dtypes
cuantas = len(variables)
for i in range(cuantas):  # variable in variables:
    variable = variables[i]
    tipo = str(tipos[i])
    if ('int' in tipo) or ('float' in tipo) :
        salida.write(str(i+1)+". Variable: "+str(variable)+'\n')
        salida.write("\tCuenta: "+str(len(df[variable]))+'\n')

        cuantosNulos =  df[variable].isnull().sum()
        if cuantosNulos == 0:
            salida.write("\tSin valores nulos\n")
        else:
            salida.write("\tValores nulos: "+str(cuantosNulos)+'\n')

        salida.write("\tMínimo: "+str(df[variable].min())+'\n')
        salida.write("\tMáximo: "+str(df[variable].max())+'\n')
        salida.write("\tPromedio: "+str(df[variable].mean(skipna = True))+'\n')
        salida.write("\tMediana: "+str(df[variable].median(skipna = True))+'\n')

        moda = df[variable].mode(dropna = True)
        valoresModa = moda.tolist()
        if len(valoresModa) == 0:
            laModa = "no hay"
        elif len(valoresModa) == 1:
            laModa = "Uni="+str(valoresModa[0])
        elif len(valoresModa) == 2:
            laModa = "Bi="+str(valoresModa[0])+","+str(valoresModa[1])
        else:
            laModa = "Multi="+str(valoresModa[0])+"..."+str(valoresModa[-1])
        salida.write("\tModa: "+str(laModa)+'\n')
        salida.write("\tDesv. estándar: "+str(df[variable].std(skipna = True))+'\n')
        salida.write("\tSesgo: "+str(df[variable].skew(skipna = True))+'\n')
        salida.write("\tCurtosis: "+str(df[variable].kurtosis(skipna = True))+'\n')
        salida.write('\n')

        med = df[variable].mean(skipna = True)
        des = df[variable].std(skipna = True)

        # OBTENCIÓN DE UN DATAFRAME PARA GRAFICAR LA DISTRIBUCIÓN DE FRECUENCIAS
        dfmuestra = pd.DataFrame()
        dfmuestra[variable] = df[variable]
        # VISUALIZA DISTRIBUCION DE FRECUENCIAS
        plt.figure(figsize=(10, 6))
        #sns.histplot(data=dfmuestra, x=variable, kde=True, color='skyblue', stat='density', linewidth=0)
        sns.histplot(data=dfmuestra, x=variable, bins=20, kde=True, color='skyblue', stat='probability', linewidth=0)
        # APROXIMACIÓN A UNA DISTRIBUCIÓN NORMAL
        xmin, xmax = plt.xlim()
        x = np.linspace(xmin, xmax, int((xmax-xmin)/10))
        p = norm.pdf(x, med, des)
        plt.plot(x, p, 'k', linewidth=2, label='Aproximación a Distribución Normal')
        plt.title('Distribución de Frecuencias y Aproximación a su Distribución Normal')
        plt.xlabel("Valores de "+str(variable))
        plt.ylabel('Frecuencia')
        plt.legend()
        plt.grid(True)
        nomVar = variable.replace("/", "")
        plt.savefig(directorio+"\\F"+nomVar+".png", dpi=150, bbox_inches='tight')
        plt.close()
        print(str(int((i+1)/cuantas*100))+'% ('+str(i+1)+") Variable=",variable)
        #plt.show()

        fig, ax = plt.subplots()
        ax.boxplot(df[variable], vert=False)
        ax.grid(True, linestyle='--', alpha=0.6)
        plt.title('Box Plot de '+variable)
        plt.savefig(directorio+"\\Box"+nomVar+".png", dpi=150, bbox_inches='tight')
        plt.close()
        #plt.show()

print()
print("Resultado en "+nomArchSalida)
salida.close()
```
